```python
import pandas as pd
import numpy as np
import  matplotlib.pyplot as plt
%matplotlib inline
```

```python
df=pd.read_csv("homeprices.csv")
df
```

```python
plt.xlabel("area")
plt.ylabel("price")
plt.scatter(df.area,df.price,color="red",marker="+")
```

```python
reg=linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict([[3300]])
```
